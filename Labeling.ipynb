```python
import pandas as pd

df = pd.read_csv('/content/drive/MyDrive/data analyst /Pertemuan 26/scraped_Petualangan_Jokowi.csv')   # ganti dengan file Anda
print('Dimensi data:', df.shape)
df.head()
```

```python
positif = [
    'bagus', 'mantap', 'puas', 'keren', 'kerennn', 'top', 'oke', 'baik', 'memuaskan',
    'nyaman', 'suka', 'love', 'best', 'hebat', 'terbaik', 'sukses', 'lancar', 'aman',
    'seru', 'kocak', 'lucu', 'menghibur', 'kreatif', 'salut', 'asyik', 'gg', 'gokil',
    'tamat', 'nagih', 'karya anak bangsa', 'bangga'
]

negatif = [
    'buruk', 'jelek', 'kecewa', 'mengecewakan', 'parah', 'kapok', 'sampah', 'norak', 'najis',
    'lambat', 'lama', 'lemot', 'rusak', 'error', 'hilang', 'ilang', 'bug', 'ngebug','susah',
    'ngelag', 'lag', 'kaku', 'delay', 'iklan', 'bosen', 'bosan', 'macet','berat', 'kurang',
    'patah', 'nyangkut','bintang 1', 'bintang satu', 'uninstal', 'hapus', 'copot', 'burik','jmbt'
]

print('Jumlah kata positif (Game):', len(positif))
print('Jumlah kata negatif (Game):', len(negatif))
```

```python
def auto_label(teks):
    if not isinstance(teks, str):
        return 'netral'
    t = teks.lower()
    skor_pos = sum(1 for k in positif if k in t)
    skor_neg = sum(1 for k in negatif if k in t)

    if skor_pos > skor_neg:
        return 'positif'
    elif skor_neg > skor_pos:
        return 'negatif'
    else:
        return 'netral'

df['sentimen_auto'] = df['teks'].apply(auto_label)
print(df['sentimen_auto'].value_counts())
df.head(10)
```
