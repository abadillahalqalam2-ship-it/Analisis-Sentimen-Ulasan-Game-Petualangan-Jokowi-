```python
!pip install google-play-scraper
from datetime import date
from google_play_scraper import reviews, Sort
import pandas as pd

# Contoh: aplikasi Gojek di Indonesia
result, _ = reviews(
    'com.rio.ahok',          # app ID dari URL Play Store
    lang='id',
    country='id',
    sort=Sort.MOST_RELEVANT,
    count=5000
)
# Filter rentang tanggal
Start_date = date(2020,3,1)
end_date = date(2026,8,1)

data = [{
    'id': r['reviewId'],
    'teks': r['content'],
    'tanggal': r['at'],
    'sumber': 'playstore'
} for r in result]

df = pd.DataFrame(data)
df.to_csv('scraped_Petualangan_Jokowi.csv', index=False)
print('Total review:', len(df))
print('Rentang Tanggal:', df['tanggal'].min(), 'sampai', df['tanggal'].max())
df.head()
```
